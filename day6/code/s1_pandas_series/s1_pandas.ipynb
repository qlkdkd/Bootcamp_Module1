```python
import pandas as pd 
count = pd.Series(
    [12, 5, 3], 
    index=["person", "car", "dog"]
) 

print(count) 
print(count["car"]) # 5 ← 이름표로 찾기 
print(count[count > 4]) # person 12, car 5 ← NumPy 와 같은 조건 고르기 print(count * 2) # 칸마다 계산도 그대로 
print(count.sum()) # 20 
print(count.to_numpy()) # [12 5 3] ← 이름표를 떼면 NumPy 배열
```

```python
import pandas as pd 
mon = pd.Series({"person": 12, "car": 5, "dog": 3}) 
tue = pd.Series({"car": 7, "person": 10, "bike": 2}) 
display(mon + tue) 
# bike NaN ← 월요일엔 없음 
# car 12.0 ← 5 + 7 
# dog NaN ← 화요일엔 없음 
# person 22.0 ← 12 + 10 
display(mon.add(tue, fill_value=0)) # 짝이 없으면 0 으로 보고 더하기 
```
